```python
import json
import pandas as pd

# Path to the extracted Mordor JSON file
file_path = "data/empire_mimikatz_logonpasswords_2020-08-07103224.json"
try:
    data = []
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                data.append(json.loads(line))
    
    df = pd.DataFrame(data)
    print(f"✅ Successfully loaded {len(df)} security events!")
    print("\nSample Columns:")
    print(df.columns.tolist()[:8])
except Exception as e:
    print("❌ Error:", e)
```
